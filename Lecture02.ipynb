# Operators

Operators are special symbols which perform operations on operands.


### Types of Operators

1. Arithmetic Operators
2. Assignment Operators
3. Comparision Operators
4. Logical Operators
5. Bitwise Operators
6. Membership Operators
7. identity Operators

### Arithmetic Operators

|Operator | Name | Example |
|-------|-------|--------|
| + |Addition | 10 + 3|
| - |Subtraction | 10 - 3 |
| * |Multiplication | 19 * 3 |
| / |Division | 20/7 |
| // |Floor Division |20/7|
| % |Modulus | 12/5 |
| ** |Exponentiation | 2 ** 4 |


```python
num1 = 34
num2 = 5

print("Addition:",num1 + num2 )
print("Subtraction:", num1 - num2)
print("Multiplication",num1 * num2)
print("Division:",num1 / num2)
print("Floor Division:",num1 // num2)
print("Remainder:",num1 % num2)
print("Exponentiation:",num2 ** 3)
```

```python
# if we use the + addition symbol with strings then it concatenate it.

s1 = "hello "
s2 = "how are you ?"

print(s1 + s2)
```

```python
print(s1 * 5)
```

### Comparision Operators


|Operator | Meaning |
|------|-------|
| == | Equal to |
| != | not equal to |
| > | Greater than |
| < | Less than |
| >= | Greater than or equal to |
| <= | Less than or equal to | 


Comaprision Operators either return True or False

```python
num1 = 45
num2 = 34

print(num1 == num2)
print(num1 != num2)
print(num1 > num2)
print(num1 < num2)
print(num1 >= num2)
print(num1 <= num2)
```

### Logical Operators

|Operator | Meaning |
|------|------|
| and | Both conditions |
| or | At least one condition |
| not | Reverse the boolean result |

```python
age =23
salary = 34000

print(age > 24 and salary > 30000)
print(age > 24 or salary > 30000)
print(not (age > 24 or salary > 30000))
```

### Bitwise Operators

|Operator | Meaning |
|------|------|
| & | Bitwise AND |
| Pipe symbol | Bitwise OR |
| ^ | XOR |
| ~ | not |
|<< | left shift |
| >> | right shift |

```python
print(10 & 12)
print(10 | 12)

print(~13) # ~n = -(n+1)

print(10 ^ 12) #(~A & B) | (A & ~B)
```

```python
print(10 << 2)
```
