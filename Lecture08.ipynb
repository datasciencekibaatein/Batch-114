## Arbitrary Positional Arguments

What if we don't know how many arguments the user will pass ?

```python
def addition(*args):
    print(args)
    print(type(args))
```

```python
addition(10,20,30,40)
```

```python
def addition(*numbers):
    total = 0

    for n in numbers:
        total += n

    return total
```

```python
addition(23,20,50,30,90,43)
```

```python
def student_marks(name,*marks):
    print("Student: ",name)
    ttl = 0
    for mark in marks:
        ttl += mark

    print("Total: ",ttl)
    print("Avg: ",ttl / len(marks))
```

```python
student_marks("Rohan",45,34,89,67,20,39,48)
```

## Arbitrary Keyword arguments

What if we want to accept an unkown number of keyword arguments ?

```python
def student_details(**kwargs):
    print(kwargs)
    print(type(kwargs))
```

```python
student_details(name = "Rahul",age = 24,course = "DS",city = "Delhi")
```

```python
def student_details(**kwargs):
    for v in kwargs.values():
        print(v, end=" ")
```

```python
student_details(name = "Rahul",age = 24,course = "DS",city = "Delhi")
```

```python
def student_details(**kwargs):
    for k in kwargs.keys():
        print(k, end=" ")
```

```python
student_details(name = "Rahul",age = 24,course = "DS",city = "Delhi")
```

```python
def student_details(**kwargs):
    for it in kwargs.items():
        print(it, end=" ")
```

```python
student_details(name = "Rahul",age = 24,course = "DS",city = "Delhi")
```

```python
def student_details(**kwargs):
    for k,v in kwargs.items():
        print(k,"=",v, end=" ")
```

```python
student_details(name = "Rahul",age = 24,course = "DS",city = "Delhi")
```

## Global And Local Scope

```python
msg = "Hello"

def greet():
    name = "Krish"
    print("hello ",name)

```

```python
greet()
```

```python
msg
```

```python
count = 10

def update_count():
    global count
    count = 5
```

```python
update_count()
```

```python
count
```

## Nesting of functions

```python
def outer():
    print("outer function")
    def inner():
        print("inner function")
    
    inner()
    print("outer after inner")

    
```

```python
outer()
```

## lambda function

A lambda function is a small anonymous function defined using the lambda keyword.

```python
def sqr(num):
    return num ** 2
```

```python
val = sqr(4)

print(val)
```

```python
s = lambda num: num ** 2
```

```python
s(6)
```
