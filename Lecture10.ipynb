## Exception

An exception is an error that occurs during program execution and interrupts the normal flow of the program.

```python
a = 10
b = 0

print(f"result: {a/b}")
print("Program Completed")
```

## What is Exception Handling ?



Exception handling is a mechanism that allows us to detect and handle runtime exception so that the program can continue executing when appropriate.


Python provides four important keywords:

- try
- catch
- else
- finally

```python
a = 234
b = 0

try:
    print(f"a / b: {a / b}")
except:
    print("Something went wrong")

    
print("Program Completed")
```

## Handling Specific Exeptions

```python
a = 234
b = 0

try:
    print(f"a / b: {a / b}")
except ZeroDivisionError:
    print("Something went wrong")

    
print("Program Completed")
```

```python
try:
    num1 = int(input("Enter the number1: "))
    num2 = int(input("Enter the number 2: "))

    print(num1/num2)
    print("Result: ",num1/num2)


except ValueError:
    print("It needed Integer Value")

except ZeroDivisionError:
    print("Divison by zero is not allowed")

except TypeError:
    print("Invalid data types")
```

## try-except with else

```python
try:
    num1 = int(input("Enter the number1: "))
    num2 = int(input("Enter the number 2: "))

    res = num1/num2

except ValueError:
    print("It needed Integer Value")

except ZeroDivisionError:
    print("Divison by zero is not allowed")

except TypeError:
    print("Invalid data types")

else:
    print("Result: ",res)
```

else; it helps us to separate normal execution logic from error-handlng logic.

## try-except : finally

```python
try:
    num1 = int(input("Enter the number1: "))
    num2 = int(input("Enter the number 2: "))

    res = num1/num2

except ZeroDivisionError:
    print("Divison by zero is not allowed")

except TypeError:
    print("Invalid data types")

else:
    print("Result: ",res)


finally:
    print("Program completed")
```
