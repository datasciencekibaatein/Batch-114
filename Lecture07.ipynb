# Functions in python

A function is a reusable block of code designed to perform a specific task. Instead of writing the same code repeatedly. We define a function once and call it whenver required.

```python
# Sum of two numbers

num1 = 34
num2 = 45
print(num1 + num2)
```

```python
num1 = 78
num2 = 40
print(num1 + num2)
```

```python
def addition(num1,num2):
    print(num1 + num2)
```

```python
addition(45,78)
```

```python
addition(90,36)
```

```python
addition(78,45)
```

## Advantages of Functions

- Code Reusability.
- Reduce code duplication.
- Better code organization.
- Easier Debugging and testing.
- Improve readability
- Supports modular programming.

## Functions with parameters and arguments

```python
def add10(num):
    print(num + 10)
```

```python
add10(34) #add10(arguments)
```

```python
# multiple parameters and arguments
def addition(num1,num2):
    print("num1: ",num1, "num2:",num2)
    print(num1 + num2)
```

```python
n1 = int(input("Enter the first number: "))
n2 = int(input("Enter the second number: "))


addition(n2,n1) # multiple arguments(positional argument)
```

```python
# Positional argument
## Arguments are assigend to parameters based on position.

def student(name,age):
    print("My nane is",name,"and my age is",age,".")
```

```python
student("Krish",24)
```

```python
student(24,"Krish")
```

```python
# Keyword arguemnt

# We can explicitly mention parameter names while calling a function.

def student(name,age,course):
    print("My name is ",name,"and age is ",age,"I'm enrolled in",course,".")
```

```python
student("krish",23,"DS")
```

```python
student(23,"krish","DS")
```

```python
student(age = 23,name = "krish",course = "DS")
```

```python
# Default Arguments


def student(name,age,course="DS"):
    print("My name is ",name,"and age is ",age,"I'm enrolled in",course,".")
```

```python
student("Rahul",26,"DE")
```

```python
student(age=26,name="Rahul",course="DE")
```

```python
student("Rahul",24,"DE")
```

#### Rule: Non-default parameters must come before default

# Return Statement

return statement sends a result back to the location where the function was called.

```python
def addition(num1,num2):
    print(num1 + num2)
```

```python
res = addition(45,35)
```

```python
print(res)
```

```python
def addition(num1,num2):
    return num1 + num2
```

```python
addition(43,78)
```

```python
res = addition(23,89)
```

```python
print(res )
```

```python
## return multiple values

def calculate(num1,num2):
    addition = num1 + num2
    subtraction = num1 - num2
    multiplication = num1 * num2

    return addition,subtraction,multiplication

```

```python
x,y,z = calculate(64,30)

print("Addition: ",x)
print("Subtraction: ",y)
print("Multiplication: ",z)
```
