### while 

```python

i = 1

while i < 11:
    print(i)
    i = i + 1
```

```python
password = "jfs423jdsf"
pwd = ""

while pwd != password:
    pwd = input("Enter the password: ")

print("Access granted")
```

#### Create the program.. keep adding the number until user enter 0

```python
total = 0

num = int(input("Enter the number: "))
while num != 0:
    total = total + num #total+=num
    num = int(input("Enter the number again: "))

print("Total is :",total)
```

### for loop

for loop works with sequence

print the numbers from 1 to 10

- range(stop) # starting with 0
- range(start,stop)
- range(start,stop,step)

in range stop will be excluded

```python
for i in range(10):
    print(i)
```

```python
for i in range(11,20):
    print(i)
```

```python
for i in range(2,21,3):
    print(i)
```

```python
# 5 - > 1

for i in range(5,0,-1):
    print(i)
```

```python
lst = [3,4,2,5,7,3,7,3]
```

```python
for i in lst:
    print(i)
```

```python
for s in "rohit":
    print(s)
```

### other statments: break,continue,pass

```python
#break is used to break the loop on given condition
for i in range(1,10):
    if i == 5:
        break
    
    print(i)
```

```python
password = "jfs423jdsf"
pwd = ""
mxAttempt = 5
ma = 1
while pwd != password:
    if ma==mxAttempt:
        break

    pwd = input("Enter the password: ")
    ma+=1

print("Access granted")
```

```python
#continue statment is used to skip the iteration on given condition
for i in range(1,10):
    if i == 5:
        continue
    
    print(i)
```

```python
#pass statment is used to do nothing
for i in range(1,10):
    if i == 5:
        pass
    
    print("hey",i)
```
