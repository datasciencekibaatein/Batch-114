1. Student Marks Analyzer 

Write a program that accepts marks of 10 students in a list and: 
- Calculate the average marks. 
- Find the highest and lowest marks. 
- Count students who passed and failed. 
- Display the grade of each student based on the marks. 

```python
std_lst = []

i = 0
while i < 10:
    marks = int(input(f"Enter the marks of student {i+1}: "))
    std_lst.append(marks)
    i = i +1

print(std_lst)

#01

ttl = 0 
for marks in std_lst:
    ttl += marks

print("Average Marks: ",ttl/len(std_lst))

#02

low = 9999
high = -1
for marks in std_lst:
    if marks > high:
        high = marks

    if marks < low:
        low = marks

print("High: ",high,"Low: ",low)

#03

# 33 > paass otherwise fail
pass_st = 0
fail = 0


for marks in std_lst:
    if marks< 33:
        fail+=1
    else:
        pass_st+=1


print("Pass: ",pass_st, "Fail: ",fail)


#04
print("-"*10)
print("Grade for marks: ")
print("-"*10)
for marks in std_lst:
    if marks>90:
        print(f"{marks} = A+")
    elif marks > 80:
        print(f"{marks} = A")
    elif marks > 70:
        print(f"{marks} = B+")
    elif marks > 60:
        print(f"{marks} = B")
    elif marks > 50:
        print(f"{marks} = C")
    elif marks >= 33:
        print(f"{marks} = D")
    else:
        print(f"{marks} = Fail")
        


```
