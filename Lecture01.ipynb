## Variable 
### Variable is the name of memory location where we store our value/data.

Rules for naming variable:

- variable should not start with any number
- it can not have any special characters
- it can not have space between, if the variable name has more than one word, instead use camel or snake case
- it should have meaning,(which is for understanding purpose)

```python
4thr = "hello"

print(4thr)
```
