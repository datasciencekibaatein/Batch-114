# Control Statement

These are the statements which controls the flow of programming.

- Conditional Statement
- Iterative Statement

if, if-else, if-elif-else, and nested if-else.

```python
num = int(input("Enter the number: "))
if num % 2 == 0:
    print("even") #indentation
else:
    print("odd")
```

if for multiple conditions

```python
age = 27
salary = 45000

if age > 20 and salary > 40000:
    print("Allowed")

else:
    print("not allowed")
```
