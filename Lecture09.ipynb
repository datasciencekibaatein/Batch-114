### map() function

The map() function applies a function to every element of an iterable.

```python
numbers = list(range(1,11))
```

```python
res = map(lambda x:x*2,numbers)
print(res)
print(list(res))
```

```python
res = []
val = lambda x:x*2 

for i in numbers:
    res.append(val(i))
```

```python
res
```

```python
val = lambda x:x*2 

```

```python
val(3)
```

### filter() function

filter() function is same like map function used to select the element from collection based on the condition.

```python
numbers
```

```python
for n in numbers:
    print(n if n%2==0 else "" ,end=" ")
```

```python
list(filter(lambda x:x%2==0,numbers))
```

### reduce
The reduce() function repeatedly applies a function to two elements, reducing iterable to single element.

```python
numbers
```

```python
from functools import reduce

res = reduce(lambda x,y:x+y,numbers)

print(res)
```

## Properties of strings

- Strings are ordered sequences.
- Strings support indexing and slicing
- Strings are immutable.
- String can contain, letters, numbers,symbols and space.

#### String indexing.

```python
topic = "Python"
print(topic)
print(len(topic))
```

```python
# Positive indexing
print(topic[0])
print(topic[1])
print(topic[2])
print(topic[3])
print(topic[4])
print(topic[5])
```

```python
# Negative Indexing

print(topic[-6])
print(topic[-5])
print(topic[-4])
print(topic[-3])
print(topic[-2])
print(topic[-1])

```
