### map() function

The map() function applies a function to every element of an iterable.

```python
numbers = list(range(1,11))
```

```python
res = map(lambda x:x*2,numbers)
print(res)
print(list(res))
```

```python
res = []
val = lambda x:x*2 

for i in numbers:
    res.append(val(i))
```

```python
res
```

```python
val = lambda x:x*2 

```

```python
val(3)
```

### filter() function

filter() function is same like map function used to select the element from collection based on the condition.

```python
numbers
```

```python
for n in numbers:
    print(n if n%2==0 else "" ,end=" ")
```

```python
list(filter(lambda x:x%2==0,numbers))
```

### reduce
The reduce() function repeatedly applies a function to two elements, reducing iterable to single element.

```python
numbers
```

```python
from functools import reduce

res = reduce(lambda x,y:x+y,numbers)

print(res)
```

### Removing space and characters

- strip() : Remove the leading and trailing characters
- lstrip() : Remove characters from the left
- rstrip() : Remove characters from the right

### Searching Functions

- find(): Returns the first index of substring
- rfind(): Returns the last index of a substring
- index(): Return the first index; raise error if missing
- rindex(): Return the last index; raise error if missing
- count(): Counts occurrences of substring

```python
text = "Python is easy to learn easy."

print("find: ",text.find("easy"))

print("rfind: ",text.rfind("not"))

print("index: ",text.index("learn"))

print("rindex: ",text.rindex("easy"))

print("count: ",text.count('easy'))
```

### Checking functions

- startswith()
- endswith()
- isalpha()
- isdigit()
- isnumeric()
- isalnum()
- isspace()
- islower()
- isupper()
- istitle()

```python
email = "example43@gmail.com"

print("startswith(): ",email.startswith("ex"))
print("endswith(): ",email.endswith("com"))

print("isalpha(): ","Python Programming".isalpha())
print("isnumeric(): ","34324".isnumeric())
print("isalnum(): ",email.isalnum())

print("isspace(): ",email.isspace())
print("islower(): ",email.islower())
print("isupper(): ",email.isupper())
print("title(): ",email.istitle())


print("12345".isnumeric())     # True
print("①②③".isnumeric())       # True (circled numbers)
print("⅓⅔".isnumeric())         # True (fractions)
print("五".isnumeric())  

print("12345".isdigit())     # True
print("①②③".isdigit())       # True (circled numbers)
print("⅓⅔".isdigit())         # True (fractions)
print("五".isdigit())

print("1/2".isnumeric())
print("453243".isdigit())

```
